# EDA on jokes

```python
text_list = []
with open('../data/jokes_clean.txt', 'r') as f:
    for line in f:
        text_list.append(str(line))
```

```python
len(text_list)
```

```python
total_symbols = 0
total_words = 0
wors_statistics = []
for line in text_list:
    total_symbols += len(line)
    total_words += len(line.split())
    wors_statistics.append(len(line.split()))
av_len_symb = total_symbols/len(text_list)
av_len_word = total_words/len(text_list)

print('Total number of symbols:', total_symbols)
print('Total number of words:', total_words)
print('Average number of symbols:', round(av_len_symb,2))
print('Average number of word:', round(av_len_word,2))
```

```python
import matplotlib.pyplot as plt

plt.subplots(2,1,figsize=(10,5))

plt.subplot(2,1,1)
plt.title('Histogram of word count')
plt.hist(wors_statistics, 152)
plt.grid(True)

plt.subplot(2,1,2)
plt.boxplot(wors_statistics, vert=False)
plt.grid(True)
plt.show()
```
