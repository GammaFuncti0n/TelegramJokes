# F3

```python
df = pd.read_parquet('../data/jokes_f3.parquet')
```

```python
df['text'] = df['conversations'].apply(lambda s: s[1]['value'])

with open('../data/jokes_txt_f3.txt', 'w',  encoding="utf-8") as f:
    for s in df['text']:
        f.write(s.replace("\n", "\\n")+"\n")
```

# Merging

```python
text_list = []
with open('../data/jokes_txt_f1.txt', 'r') as f:
    for line in f:
        text_list.append(str(line))

with open('../data/jokes_txt_f2.txt', 'r') as f:
    for line in f:
        text_list.append(str(line))

with open('../data/jokes_txt_f3.txt', 'r') as f:
    for line in f:
        text_list.append(str(line))
```

```python
with open('../data/jokes_merged.txt', 'w',  encoding="utf-8") as f:
    for s in text_list:
        f.write(s.strip()+"\n")
```
